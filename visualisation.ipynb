```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

## Handlers 

```python
def find_csv_path(folder_path):
    files = os.listdir(folder_path)
    for file in files:
        if file.endswith('.csv'):
            return os.path.join(folder_path, file)
```

```python
# Null values in the dataset
null_values = covid_data_df.isnull().sum()
null_values[null_values > 0]
```

```python
# The initial dataset has dates as columns, we need to transform this dataframe to have
# 'Country/Region', 'Date' and 'Cases' columns for easier plotting.
df_melted = covid_data_df.melt(id_vars=['Country/Region', 'Province/State', 'Lat', 'Long'], 
                            var_name='Date', 
                            value_name='Cases')

# Convert the 'Date' column to datetime
df_melted['Date'] = pd.to_datetime(df_melted['Date'], format='%m/%d/%y')

# We will now group by country and date to get the total number of cases for each country on each date.
df_grouped = df_melted.groupby(['Country/Region', df_melted['Date'].dt.to_period('M')])['Cases'].max().reset_index()

# Convert the 'Date' from period to datetime to plot
df_grouped['Date'] = df_grouped['Date'].dt.to_timestamp()

# Now we pivot the table to have countries as columns and dates as rows for plotting
df_pivot = df_grouped.pivot(index='Date', columns='Country/Region', values='Cases').ffill()

# Determine the number of countries to include in each plot
total_countries = len(df_pivot.columns)
countries_per_plot = total_countries // 10
if total_countries % 10:  # If there's a remainder, add one more to ensure all countries are included
    countries_per_plot += 1

# Initialize a list to store the figures for later display
figures = []

# Create 10 figures, each with a subset of countries
for i in range(0, total_countries, countries_per_plot):
    fig, ax = plt.subplots(figsize=(15, 5))
    subset = df_pivot.columns[i:i + countries_per_plot]
    df_pivot[subset].plot(ax=ax)
    ax.set_title(f'COVID-19 Confirmed Cases: Countries {i+1} to {min(i + countries_per_plot, total_countries)}')
    ax.set_xlabel('Date')
    ax.set_ylabel('Number of Confirmed Cases')
    ax.legend(title='Country/Region', bbox_to_anchor=(1.05, 1), loc='upper left')
    ax.grid(True)
    plt.xticks(rotation=90)
    plt.tight_layout()
    figures.append(fig)

# Display the first figure as an example
plt.figure(figures[0].number)
plt.show()

# Save the figures to PNG files
for i, fig in enumerate(figures):
    fig.savefig(f'results/figures/covid-19-confirmed-cases-{i+1}.png', bbox_inches='tight')

```

## Query 1

```python
query1_df = pd.read_csv(find_csv_path("results/query1/mean_daily_cases_per_month_spark"))
query1_df.head()
```

```python
pivot_df = query1_df.pivot_table(index=['Year', 'Month'], columns='Country/Region', values='Average')
print(pivot_df.shape)
pivot_df.head()
```

```python
# Calculate the number of countries for each third
countries_per_figure = len(pivot_df.columns) // 3
if len(pivot_df.columns) % 3:  # If there's a remainder, add one more plot to the last figure
    countries_per_figure += 1

# Create a list of dataframes, each with a subset of countries
sub_dfs = [pivot_df.iloc[:, i:i + countries_per_figure] for i in range(0, len(pivot_df.columns), countries_per_figure)]

# Plot each subset on a separate figure
for i, df_subset in enumerate(sub_dfs):
    fig, ax = plt.subplots(figsize=(18, 8))
    sns.heatmap(df_subset, cmap="YlGnBu", ax=ax)
    ax.set_title(f'Mean number of daily confirmed cases for each month - Countries {i*countries_per_figure+1} to {min((i+1)*countries_per_figure, len(pivot_df.columns))}')
    ax.set_xlabel('Country/Region')
    ax.set_ylabel('Date (Year-Month)')
    ax.tick_params(axis='x', rotation=90)
    ax.tick_params(axis='y', rotation=0)
    plt.tight_layout()
    plt.show()

    # Save the figure to a PNG file
    fig.savefig(f'results/figures/mean-daily-confirmed-cases-per-month-{i+1}.png', bbox_inches='tight')
```

## Query 2

### Top 100 Locations 

```python
top100Locations = pd.read_csv(find_csv_path("results/query2/top_100_affected"))

# Keep only the location and MeanSlope columns
top100Locations = top100Locations[['Location', 'MaxSlope']].drop_duplicates()

# Order the locations by their mean slope
top100Locations = top100Locations.sort_values(by='MaxSlope', ascending=False)

top100Locations.head()
```

```python
# Add the Location column to the CovidData DataFrame : Province/State or Country/Region if Province/State is null
covid_data_df["Location"] = covid_data_df["Province/State"].fillna(covid_data_df["Country/Region"])

# Add the Long and Lat columns to the the Query3 DataFrame by merging with the CovidData DataFrame
top100Locations = pd.merge(top100Locations, covid_data_df[["Location", "Long", "Lat"]], on="Location")

top100Locations.head()
```

```python
import geopandas as gpd

# Read the shapefile
world = gpd.read_file('data/World_Continents.zip')

# Convert the DataFrame to a GeoDataFrame
gdf = gpd.GeoDataFrame(top100Locations, geometry=gpd.points_from_xy(top100Locations.Long, top100Locations.Lat))

# Plotting
fig, ax = plt.subplots(figsize=(20, 10))
world.boundary.plot(ax=ax, linewidth=1)

# Define a scale factor for the circle sizes
scale_factor = 0.25

for idx, row in gdf.iterrows():
    ax.scatter(row['Long'], row['Lat'], s=row['MaxSlope'] * scale_factor, alpha=0.6)

plt.title('Top 100 locations most affected by the pandemic')
plt.xlabel('Longitude')
plt.ylabel('Latitude')
plt.grid(True)
plt.tight_layout()
plt.show()

# Save the figure to a PNG file
fig.savefig('results/figures/top-100-locations-most-affected.png', bbox_inches='tight')
```

```python
query2_df = pd.read_csv(find_csv_path("results/query2/weekly_stats_spark"))
query2_df.head()
```

```python
# Create a new column "WeekStart" to store the start date of the week
query2_df['WeekStart'] = pd.to_datetime(query2_df['WeekRange'].apply(lambda x: x.split(' - ')[0]))

# Now let's sort the DataFrame by 'Continent' and 'WeekStart'
df_uploaded_sorted = query2_df.sort_values(by=['Continent', 'WeekStart'])

# Create a new column "WeekNumber" to store the week number
df_uploaded_sorted['WeekNumber'] = df_uploaded_sorted.groupby('Continent').cumcount() + 1

# Since the data is now sorted, we can drop the 'WeekStart' as it was only needed for sorting
df_uploaded_sorted.drop('WeekStart', axis=1, inplace=True)

# Reset index for plotting
df_uploaded_sorted.reset_index(drop=True, inplace=True)

# Plotting the mean number of confirmed cases by week for each continent
figure = plt.figure(figsize=(14, 7))
sns.lineplot(data=df_uploaded_sorted, x='WeekNumber', y='Mean', hue='Continent', marker='o')

# Improve readability of the plot
plt.xticks(rotation=90)
plt.title('Mean Number of Confirmed COVID-19 Cases by Week and Continent')
plt.xlabel('Week Number')
plt.ylabel('Mean Number of Confirmed Cases')
plt.legend(title='Continent', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Show the plot
plt.show()

# Save the figure to a PNG file
figure.savefig('results/figures/mean-confirmed-cases-by-week-and-continent.png', bbox_inches='tight')
```

```python
# Plotting the min number of confirmed cases by week for each continent
figure = plt.figure(figsize=(14, 7))
sns.lineplot(data=df_uploaded_sorted, x='WeekNumber', y='Min', hue='Continent', marker='o')

# Improve readability of the plot
plt.xticks(rotation=90)
plt.title('Min Number of Confirmed COVID-19 Cases by Week and Continent')
plt.xlabel('Week Number')
plt.ylabel('Min Number of Confirmed Cases')
plt.legend(title='Continent', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Show the plot
plt.show()

# Save the figure to a PNG file
figure.savefig('results/figures/min-confirmed-cases-by-week-and-continent.png', bbox_inches='tight')
```

```python
# Plotting the max number of confirmed cases by week for each continent
figure = plt.figure(figsize=(14, 7))
sns.lineplot(data=df_uploaded_sorted, x='WeekNumber', y='Max', hue='Continent', marker='o')

# Improve readability of the plot
plt.xticks(rotation=90)
plt.title('Max Number of Confirmed COVID-19 Cases by Week and Continent')
plt.xlabel('Week Number')
plt.ylabel('Max Number of Confirmed Cases')
plt.legend(title='Continent', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()

# Show the plot
plt.show()

# Save the figure to a PNG file
figure.savefig('results/figures/max-confirmed-cases-by-week-and-continent.png', bbox_inches='tight')
```
